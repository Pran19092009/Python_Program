**Solving Missing values Problem using the fillna(), ffill(),replace()**

```python
email[email["Department"].isna()]
```

```python
email["Department"]=email["Department"].fillna("IT")
```

```python
email["Region"]=email["Region"].replace(to_replace=np.nan,value="East")
```

```python
email["Customer_Type"]=email["Customer_Type"].ffill()
```

**Droping some Columns when have missing data** 

```python
email.drop("Replay_within_24Hrs",axis=1,inplace=True)
```

```python
email.drop("Network_type",axis=1,inplace=True)
```

**Viualization on DataSet Using MATPLOTLIB**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
graph=email.groupby("Satisfaction_Rating")["Email_ID"].sum()
```

```python
x=graph.index
```

```python
y=graph.values
```

```python
plt.figure(figsize=(9,4),facecolor="#EFEABB")
plt.title("*Satisfaction Rating*",color="#17433F")
plt.scatter(x,y)
for i in range(len(y)):
    plt.text(x[i],y[i],y[i].astype(int))
```
