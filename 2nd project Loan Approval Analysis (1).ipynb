#Fill categorical columns

```python
loan["Marital_Status"]=loan["Marital_Status"].fillna("Single")
```

**access The Marital_Status column for fillign their NaN values then filled their values using fillna() function and fill Single where NaN persent and loaded into the original dataset**

```python
loan["Gender"]=loan["Gender"].replace(to_replace=np.nan,value="Male")
```

**access the Gender column for filling their NaN values then filled their values using replace() function between the replace function used to_replace menthod is assign equal to np.Nan and replacing equal to value " Male" string and loaded into the original dataset**

```python
loan["Credit_History"]=loan["Credit_History"].fillna("Good")
```

**Accessed the Credit_History column and replaced all missing (NaN) values with "Good" using the fillna() function. The updated values were then saved back into the original dataset.**

```python
loan["Approval_Reason"]=loan["Approval_Reason"].fillna("Pending")
```

**Accessed the Approval_Reason column and replaced all missing (NaN) values with "Pending" using the fillna() function. The updated values were then saved back into the original dataset.**

```python
# Fill numerical columns
```

```python
loan["Annual_Income"]=loan["Annual_Income"].fillna(loan["Annual_Income"].median())
```

**Accessed the Annual_Income column and calculated its median value. Then, using the fillna() function, replaced all missing (NaN) values with the median and saved the updated values back into the original dataset.**

```python
loan["Loan_Amount"]=loan["Loan_Amount"].fillna(loan["Loan_Amount"].median())
```

**Accessed the Loan_Amount column and calculate its median value .Then ,using the fillna function ,replaced all missing (NaN) values with the median and saved the update value back into the original dataset**

```python
loan["Credit_Score"]=loan["Credit_Score"].fillna(loan["Credit_Score"].median())
```

**Accessed the Credit_Score column and calculate its median value .Then ,using the fillna function ,replaced all missing (NaN) values with the median and saved the update value back into the original dataset**

```python
loan["Savings_Balance"]=loan["Savings_Balance"].fillna(loan["Savings_Balance"].median())
```

**Accessed the Savings_Balance column and calculate its median value .Then ,using the fillna function ,replaced all missing (NaN) values with the median and saved the update value back into the original dataset**
