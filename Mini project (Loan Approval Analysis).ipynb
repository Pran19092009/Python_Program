# **Risk Aanlysis**
### The analysis helps identify important factors related to loan approval, rejection, applicant risk, and loan application patterns.

```python
loan.groupby("Risk_Level")["Loan_Status"].
```

```python
loan["Loan_Status"]
```

```python
risk_status = pd.crosstab(
    loan["Risk_Level"],
    loan["Loan_Status"],
    normalize="index") * 100

risk_status.plot(kind="bar",figsize=(8,5),color=["#FFD400","#FF0052"])

plt.title("Loan Approval Status by Risk Level",
          fontsize=14, fontweight="bold")

plt.xlabel("Risk Level")
plt.ylabel("Percentage (%)")

plt.grid(axis="y", alpha=0.4)
plt.gca().set_axisbelow(True)

plt.xticks(rotation=0)
plt.legend(title="Loan Status")

plt.tight_layout()
plt.show()
```
